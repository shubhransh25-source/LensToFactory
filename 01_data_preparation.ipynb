```python
# FactoryLens | Notebook 1 — Data Preparation
# Step 1: Import libraries

import pandas as pd
import numpy as np
import os
```

```python
# Cell 2 — Load the dataset

# Define the path to your CSV file
# Replace the path below with your actual file path
data_path = r"C:\Users\yashp\Downloads\MyFactoryLens\ai4i+2020+predictive+maintenance+dataset\ai4i2020.csv"

# Check if the file exists before loading
if os.path.exists(data_path):
    print("File found. Loading dataset...")
    df = pd.read_csv(data_path)
    print("Dataset loaded successfully.")
    print(f"Shape: {df.shape[0]} rows and {df.shape[1]} columns")
else:
    print("File not found. Check your path.")
```

```python
# Cell 3 — First look at the data

# See the first 5 rows
print("FIRST 5 ROWS")
display(df.head())

# See all column names
print("\nCOLUMN NAMES ")
print(df.columns.tolist())

# See data types of each column
print("\nDATA TYPES ")
print(df.dtypes)
```

```python
# Cell 4 — Basic statistics

print("=== STATISTICAL SUMMARY ===")
display(df.describe())

print("\n=== TARGET COLUMN — Failure Distribution ===")
failure_counts = df['Machine failure'].value_counts()
failure_percent = df['Machine failure'].value_counts(normalize=True) * 100

print(f"No Failure (0): {failure_counts[0]} rows — {failure_percent[0]:.1f}%")
print(f"Failure (1):    {failure_counts[1]} rows — {failure_percent[1]:.1f}%")
```

```python
# Cell 5 — Check for missing values

print("=== MISSING VALUE CHECK ===\n")

# Count missing values per column
missing_count = df.isnull().sum()

# Calculate percentage
missing_percent = (df.isnull().sum() / len(df)) * 100

# Combine into one clean table
missing_df = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %': missing_percent
})

# Only show columns that actually have missing values
missing_df = missing_df[missing_df['Missing Count'] > 0]

if len(missing_df) == 0:
    print("✅ No missing values found in any column.")
else:
    print("⚠️ Missing values found:")
    display(missing_df)
```

```python
# Cell 6 — Check for duplicates

duplicate_count = df.duplicated().sum()

print(f"Total duplicate rows: {duplicate_count}")

if duplicate_count == 0:
    print("✅ No duplicate rows found.")
else:
    print(f"⚠️ Found {duplicate_count} duplicate rows. Dropping them.")
    df = df.drop_duplicates()
    print(f"✅ Duplicates removed. New shape: {df.shape}")
```

```python
# Cell 7 — Clean column names

print("=== BEFORE CLEANING ===")
print(df.columns.tolist())

# Rename columns to clean, professional names
df = df.rename(columns={
    'UDI': 'udi',
    'Product ID': 'product_id',
    'Type': 'product_type',
    'Air temperature [K]': 'air_temp_k',
    'Process temperature [K]': 'process_temp_k',
    'Rotational speed [rpm]': 'rotational_speed_rpm',
    'Torque [Nm]': 'torque_nm',
    'Tool wear [min]': 'tool_wear_min',
    'Machine failure': 'machine_failure',
    'TWF': 'failure_twf',
    'HDF': 'failure_hdf',
    'PWF': 'failure_pwf',
    'OSF': 'failure_osf',
    'RNF': 'failure_rnf'
})

print("\n=== AFTER CLEANING ===")
print(df.columns.tolist())

print("\n✅ Column names cleaned successfully.")
```
